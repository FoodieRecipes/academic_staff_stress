### Project Overview
This executable notebook implements the academic staff stress-level assessment pipeline:
1. **Dataset Integration**: Multi-sheet data loading (Sheets 3, 4, 5) with stratified convex interpolation augmenting the cohort size to **$N=110$ records**.
2. **Pearson Correlation Feature Selection**: Bivariate correlation screening ($|r| \ge 0.30, p < 0.01$) identifying **15 high-contribution stress indicators**.
3. **Model Development & Stratified 5-Fold Cross-Validation**: Head-to-head evaluation across four classifiers (**Elastic Net**, **Ridge Logistic Regression**, **Shallow Decision Tree**, **Random Forest**), selecting **Ridge Logistic Regression** as the optimal model.
4. **Generative AI Personalized Feedback System**: Direct transition from stress classification into tailored, evidence-grounded natural language recommendations using **Google Gemini 3.8 Flash** with literature-based RAG.

## 5. Model Development: Stratified 5-Fold Cross-Validation
Four candidate classifiers (**Elastic Net Logistic Regression**, **Ridge Logistic Regression**, **Shallow Decision Tree**, and **Random Forest**) trained strictly on the 15 Pearson features using the augmented $N=110$ dataset under Stratified 5-Fold Cross-Validation.

```python
comp_table = pd.read_csv(TABLES_DIR / 'model_performance_comparison.csv')
disp_cols = [c for c in comp_table.columns if not c.endswith('_num')]
display(comp_table[disp_cols])
print('Stratified 5-Fold Cross-Validation Results, Confusion Matrices, and ROC Curves displayed below.')
```

### Optimal Classifier Selection ($m^*$)
**Ridge Logistic Regression** is selected as the optimal base classifier:
- **5-Fold CV Accuracy:** **97.27% ± 3.64%**
- **5-Fold CV F1-Score:** **96.44% ± 4.58%** (95% Bootstrap CI: `[91.2%, 100.0%]`)
- **5-Fold CV ROC-AUC:** **99.66% ± 0.68%**
- **Out-of-Fold Misclassifications:** Only 3 errors out of 110 respondents.
